```python
from tensorflow.keras.datasets import mnist
from tensorflow.keras.utils import to_categorical

# Chargement des données
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Aplatissement des images (28x28 -> 784) et normalisation (0 à 1)
x_train = x_train.reshape(-1, 784) / 255.0
x_test = x_test.reshape(-1, 784) / 255.0

# Encodage One-Hot exigé par 'categorical_crossentropy'
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# 1. Création de la boîte vide (le modèle séquentiel)
model_softmax = Sequential()

# 2. Ajout de la couche de sortie dense de 10 neurones avec Softmax
# L'input_shape=(784,) représente ta couche d'entrée de dimension 784
model_softmax.add(Dense(10, activation='softmax', input_shape=(784,)))

# 3. Configuration du "moteur" d'apprentissage selon la consigne
model_softmax.compile(optimizer='adam',
                      loss='categorical_crossentropy',
                      metrics=['accuracy'])

# 4. Affichage de l'architecture (très utile pour noter le nombre de paramètres pour ta question e !)
model_softmax.summary()

# 5. Entraînement de l'IA (on la fait bosser pendant 10 "epochs" ou cycles)
print("--- Début de l'entraînement ---")
history_softmax = model_softmax.fit(x_train, y_train,
                                    epochs=10,
                                    validation_data=(x_test, y_test))
```

```python
# Question b. Implémenter le réseau avec 2 couches cachées et ReLU
model_relu = Sequential()

# Couche cachée 1 (ex: 128 neurones) avec fonction ReLU
model_relu.add(Dense(128, activation='relu', input_shape=(784,)))

# Couche cachée 2 (ex: 64 neurones) avec fonction ReLU
model_relu.add(Dense(64, activation='relu'))

# Couche de sortie (toujours 10 neurones et softmax pour les probabilités)
model_relu.add(Dense(10, activation='softmax'))

# Compilation (on garde les mêmes paramètres que le modèle A)
model_relu.compile(optimizer='adam',
                   loss='categorical_crossentropy',
                   metrics=['accuracy'])

# On affiche l'architecture pour voir la différence de taille !
model_relu.summary()

# Entraînement
print("--- Début de l'entraînement du Modèle ReLU ---")
history_relu = model_relu.fit(x_train, y_train,
                              epochs=10,
                              validation_data=(x_test, y_test))
```

```python
# Question c. Réseau avec 1 couche cachée et tanh
model_tanh = Sequential([
    Dense(128, activation='tanh', input_shape=(784,)),
    Dense(10, activation='softmax')
])

model_tanh.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
print("--- Début de l'entraînement du Modèle Tanh ---")
history_tanh = model_tanh.fit(x_train, y_train, epochs=10, validation_data=(x_test, y_test))
```

```python
# Question d. Réseau avec 1 couche cachée et sigmoid
model_sigmoid = Sequential([
    Dense(128, activation='sigmoid', input_shape=(784,)),
    Dense(10, activation='softmax')
])

model_sigmoid.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
print("--- Début de l'entraînement du Modèle Sigmoid ---")
history_sigmoid = model_sigmoid.fit(x_train, y_train, epochs=10, validation_data=(x_test, y_test))
```
